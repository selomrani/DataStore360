Question: Créer une nouvelle colonne 'SpendingCategory' (par ex., 'High Spender' si total > 100, 'Regular Spender' sinon) pour chaque membre, puis afficher la distribution de ces catégories par ville.

```python
data = {
    "MemberName": ["Nadia El Amrani", "Hugo Moreau", "Nadia El Amrani", "Chloe Fontaine",
                    "Marc Petit", "Hugo Moreau", "Chloe Fontaine", "Nadia El Amrani",
                    "Marc Petit", "Chloe Fontaine"],
    "City": ["Paris", "Lyon", "Paris", "Marseille",
             "Lyon", "Lyon", "Marseille", "Paris",
             "Lyon", "Marseille"],
    "Amount": [45.0, 60.0, 45.0, 30.0, 75.0, 60.0, 30.0, 45.0, 75.0, 30.0],
    "SessionType": ["Musculation", "Yoga", "Cardio", "Yoga",
                     "Musculation", "Cardio", "Musculation", "Yoga",
                     "Cardio", "Yoga"],
}

df = pd.DataFrame(data)
```

```python
client_data = df.groupby("MemberName").agg({
    "Amount" : "sum"
})
client_data
```

```python
client_data['SpendingCategory'] = np.where(client_data['Amount'] > 100 , "High Spender" , "Regular Spender")
```
