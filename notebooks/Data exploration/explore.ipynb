Investigating anomalies and outliers :

```python

df['Order Date'] = pd.to_datetime(df['Order Date'], format='mixed')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='mixed')

# Now let's check real shipping date anomalies:
real_shipping_anomalies = df[df['Ship Date'] < df['Order Date']]
print(f"Real shipping anomalies (Ship Date < Order Date): {len(real_shipping_anomalies)}")
real_shipping_anomalies[['Order ID', 'Order Date', 'Ship Date']].head()
```

Products that has negative profit ( losses) :

```python
losing_products = df[df['Profit'] < 0]
print(f"Total transactions with negative profit: {len(losing_products)}")
losing_products.head()
```

Some money / profit statistics for each category

```python
# here we can see how much each category earned for the business and how many order for each of one of those categories
categories_stats = df.groupby('Category').agg({
    'Profit': 'sum',
    'Order ID' : 'count'
})
# Todo : i noticed here that some of the categories names are duplicated , like "furniture and Furniture" , this needs to be treated later in the data cleaning phase

# 1. Top category in profit
best_category = categories_stats['Profit'].idxmax()
best_profit_val = categories_stats['Profit'].max()
print(f"Top category in profit: {best_category} (£{best_profit_val:,.2f})")

# 2. Worst category in profit
worst_category = categories_stats['Profit'].idxmin()
worst_profit_val = categories_stats['Profit'].min()
print(f"Worst category in profit: {worst_category} (£{worst_profit_val:,.2f})")

# 3. Highest selling category (order volume)
highest_order_category = categories_stats['Order ID'].idxmax()
highest_order_val = categories_stats['Order ID'].max()
print(f"Top category in order volume: {highest_order_category} ({highest_order_val:,} orders)")

# 4. Worst selling category (order volume)
worst_order_category = categories_stats['Order ID'].idxmin()
worst_order_val = categories_stats['Order ID'].min()
print(f"Worst category in order volume: {worst_order_category} ({worst_order_val:,} orders)")
```

Checking discount and quantity anomalies :

```python
# 1. Discounts > 100% (Discount > 1.0 means paying customer to take product)
crazy_discounts = df[df['Discount'] > 1.0]
print(f"Total orders with discount > 100%: {len(crazy_discounts)}")
crazy_discounts[['Order ID', 'Sales', 'Quantity', 'Discount', 'Profit']].head()
```

```python
# 2. Invalid Quantities (<= 0)
invalid_quantity = df[df['Quantity'] <= 0]
print(f"Total orders with quantity <= 0: {len(invalid_quantity)}")
invalid_quantity[['Order ID', 'Product Name', 'Quantity', 'Sales', 'Profit']].head()
```

Data Visualizations :

```python
# 1. Distribution of Sales and Profit
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

sns.histplot(df['Sales'].dropna(), bins=40, kde=True, ax=axes[0], color='skyblue')
axes[0].set_title("Distribution of Sales")
axes[0].set_xlabel("Sales ($)")

sns.histplot(df['Profit'].dropna(), bins=40, kde=True, ax=axes[1], color='salmon')
axes[1].set_title("Distribution of Profit")
axes[1].set_xlabel("Profit ($)")

plt.tight_layout()
plt.show()
```

```python
# 2. Investigating Why Profit is Negative (Discount vs Profit)
# We can see that deep discounts (> 20%) are the main driver of negative profits
plt.figure(figsize=(9, 4))
sns.scatterplot(data=df, x='Discount', y='Profit', alpha=0.5, color='darkblue')
plt.axhline(0, color='red', linestyle='--', label='Zero Profit Line')
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit ($)")
plt.legend()
plt.show()
```

```python
# 3. Distribution of Delivery Time (in days)
delivery_days = (df['Ship Date'] - df['Order Date']).dt.days
valid_delivery = delivery_days[(delivery_days >= 0) & (delivery_days <= 10)]

plt.figure(figsize=(8, 4))
sns.countplot(x=valid_delivery.astype(int), color='steelblue')
plt.title("Delivery Time Distribution (Days)")
plt.xlabel("Days to Ship")
plt.ylabel("Order Count")
plt.show()
```

```python
# 4. Correlation matrix between numerical columns
plt.figure(figsize=(7, 4))
sns.heatmap(df[['Sales', 'Quantity', 'Discount', 'Profit']].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```
